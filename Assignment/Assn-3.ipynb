### Figure 2

Use `data-fig2.npy` for this figure.

```python
import numpy as np
from matplotlib import pyplot as plt

b = np.load( 'data-fig2.npy' )
labels = ["Red", "Green", "Blue"]

plt.figure()
plt.plot(b[:, 0], color = "Red")
plt.plot(b[:, 1], color = "Green")
plt.plot(b[:, 2], color = "Blue")
plt.xlabel( "X Value" )
plt.legend( labels )
plt.ylabel( "Y Value" )
plt.show()
```

### Figure 3

Use `data-fig3.npy` for this figure.

```python
import numpy as np
from matplotlib import pyplot as plt 

c = np.load( 'data-fig3.npy' )

plt.figure()
plt.scatter(c[:, 0], c[:, 1], s=200, c= 'blue', edgecolors='face', alpha = 0.3)
plt.xlabel('X value')
plt.ylabel('Y value')
plt.show()
```

### Figure 4

Use `data-fig4.npy` for this figure.

```python
import numpy as np
from matplotlib import pyplot as plt 

d = np.load( 'data-fig4.npy' )
categories = ['A', 'B', 'C', 'D', 'E']

plt.figure()
plt.bar(categories, d, width = 0.8, color = '#7f7f7f')
plt.xlabel( 'Category' )
plt.ylabel( 'Value' )
plt.show()
```

### Figure 5

Use `data-fig5.npy` for this figure.

*Hint*: Use 2D visualization and the "RdBu" colormap.

```python
import numpy as np 
from matplotlib import pyplot as plt 

e = np.load( 'data-fig5.npy' )

plt.figure()
f = plt.pcolor ( e , cmap = 'RdBu', vmin = -1, vmax = 1)
plt.colorbar( f )
plt.show()
```
