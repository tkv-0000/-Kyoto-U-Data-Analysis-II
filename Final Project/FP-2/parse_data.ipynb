# parse_data.ipynb

This notebook parses the data files used for the FP-2 assignment. 

<br>
<br>

First let's read the attached data file:

```python
import pandas as pd

df0 = pd.read_csv( 'Raw Data_Paylater and Non Paylater User.csv' , sep = ';' )

df0.describe()
```

<br>
<br>

The dependent, independent, and control variables (DVs, IVs, and CVs) that we are interested in are:

**DVs**:
- IBB1: Respondents do most of their online shopping spontaneously
- IBB2: Respondents make purchases without planning, and they do not intend to shop when opening the application
- IBB3: Respondents have no desire to shop before opening the application
- IBB4: Respondents cannot resist shopping on the application

**IVs**:
- Promotion (column 'P1' -> 'P4' inside the csv file)
- Social Influence (column 'SI1' -> 'SI6' inside the csv file)
- Happiness (column 'H1' -> 'H4' inside the csv file)
- Self-Control (column 'SC1' -> 'SC5' inside the csv file)
- Normative Evaluation (column 'NE1' -> 'NE5' inside the csv file)

**CVs**:
- Gender: Male or Female
- E-Paylater User Status: Whether respondents have used or have never used E-Paylater services
- Monthly Income: Divided into 6 levels

<br>
<br>

Let's extract the relevant columns:

```python
df = df0[  ['IBB1', 'IBB2', 'IBB3', 'Gender', 'E-Paylater User Status', 'Monthly Income',
            'IBB4', 'P1', 'P2', 'P3', 'P4',
            'SI1', 'SI2', 'SI3', 'SI4', 'SI5', 'SI6',
            'H1', 'H2', 'H3', 'H4',
            'SC1', 'SC2', 'SC3 ', 'SC4 ', 'SC5',
            'NE1', 'NE2', 'NE3', 'NE4', 'NE5']  ]

df.describe()
```

<br>
<br>

Next let's use the `rename` function to give the columns simpler variable names:

```python
df = df.rename( columns={'E-Paylater User Status':'Status', 'Monthly Income':'Income', 'SC3 ': 'SC3', 'SC4 ':'SC4'} )   # column SC3 and SC4 have redundant spaceback in the dataset

df.describe()
```
