### Purpose:

* The main purpose of the Final project is to analyze factors influencing impulsive online buying behavior among users and non-users of e-Paylater services
* The key **dependent variable** (DV) of the dataset is Impulsive Buying Behavior (IBB):
* There are 5 **independent variables** (IVs) of the dataset are:
    - Promotion
    - Social Influence
    - Happiness
    - Self-Control
    - Normative Evaluation
* There are 3 **control variables** (CVs) of the dataset are:
    - Gender
    - E-Paylater User Status
    - Monthly Income
* The dataset has 811 cases from 811 unique respondants. All accurate DVs and IVs are present in all rows.
<br>

### Dataset source:

- The dataset gathers responses from online impulsive buying behavior of buy now pay later users and non-buy now pay later users in Indonesia and available in [ScienceDirect](https://www.sciencedirect.com/).
