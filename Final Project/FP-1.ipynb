### Purpose:

* The main purpose of the Final project is to analyze factors influencing impulsive online buying behavior among users and non-users of e-Paylater services
* The key **dependent variables** (DVs) of the dataset are Impulsive Buying Behavior (IBB):
    - IBB1
    - IBB2
    - IBB3
    - IBB4
* The key **independent variables** (IVs) of the dataset are:
    - P1 -> P4 (Promotion)
    - SI1 -> SI6 (Social Influence)
    - H1 -> H4 (Happiness)
    - SC1 -> SC5 (Self-Control)
    - NE1 -> NE5 (Normative Evaluation)
* The dataset has 811 cases from 811 unique respondants. All accurate DVs and IVs are present in all rows.
<br>

### Dataset source:

- The dataset gathers responses from online impulsive buying behavior of buy now pay later users and non-buy now pay later users in Indonesia and available in [ScienceDirect](https://www.sciencedirect.com/).

- The direct link for the dataset is available for download [here](https://data.mendeley.com/datasets/wkjdmrmrg4/4)

- The dataset originates from an article publised by Verni Juita, Vera Pujani, Rida Rahim, Rita Rahayu (2024)
